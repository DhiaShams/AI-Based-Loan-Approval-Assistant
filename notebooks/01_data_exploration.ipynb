# Data exploration for the sampled Lending Club dataset

This notebook inspects the output of the data filtering step at `data/interim/loans_sample_v1.csv`.
It is designed to run against the real local CSV and contains no fabricated charts or synthetic values.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_PATH = "../data/interim/loans_sample_v1.csv"


df = pd.read_csv(DATA_PATH)
print(f"Shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
print(df.dtypes)
print("Missing values:\n", df.isna().sum())
print("Target distribution:\n", df["loan_status"].value_counts(dropna=False))

num_cols = df.select_dtypes(include=["number"]).columns.tolist()
for col in ["loan_amnt", "annual_inc", "dti", "fico_range_low", "fico_range_high", "revol_util"]:
    if col in df.columns:
        print(f"\nDistribution for {col}:\n{df[col].describe()}")
        plt.figure(figsize=(6, 4))
        sns.histplot(df[col].dropna(), bins=30, kde=True)
        plt.title(f"{col} distribution")
        plt.tight_layout()
        plt.show()

cat_cols = df.select_dtypes(exclude=["number"]).columns.tolist()
for col in cat_cols[:10]:
    if col in df.columns:
        print(f"\n{col} distribution:\n{df[col].value_counts(dropna=False).head(10)}")

if "loan_status" in df.columns:
    default_rate = (df["loan_status"] == "Charged Off").mean()
    print(f"Default rate: {default_rate:.4f}")

corr = df[num_cols].corr(numeric_only=True)
plt.figure(figsize=(10, 8))
sns.heatmap(corr, cmap="coolwarm", annot=False)
plt.title("Feature correlation heatmap")
plt.tight_layout()
plt.show()
```
